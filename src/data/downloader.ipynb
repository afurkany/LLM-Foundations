### Request for Text File: the-verdict.txt

```python
import os
import requests

from pathlib import Path


def request_data(file_link: str, save_folder_path: Path, save_file_name):
    f_list = os.listdir(save_folder_path)

    if save_file_name not in f_list:
        response = requests.get(file_link)

        if response.status_code == 200:
            save_folder_path.joinpath(save_file_name).write_bytes(response.content)
        else:
            raise ConnectionError(f"Could not download! Error code: {response.status_code}")
    else:
        print("[INFO]: The file is already available in the folder!")

save_folder_path = Path.cwd()
file_link = r"https://raw.githubusercontent.com/rasbt/LLMs-from-scratch/refs/heads/main/ch02/01_main-chapter-code/the-verdict.txt"
request_data(file_link, save_folder_path, save_file_name="the-verdict.txt")
```
