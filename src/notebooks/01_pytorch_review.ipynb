### Tensors

Tensors generalizes vectors and matrices to higher dimensions.

* A scalar value (just a number) is a zero dimensional tensor.
* A vector is a one dimensional tensor.
* A matrix is a two dimensional tensor in PyTorch.

We can also define higher dimensional tensors like 3D tensor, and so forth which are efficiently computed in PyTorch.

Let's create some example PyTorch tensors.

```python
tensor_0d = torch.tensor(1)  # zero dimensional tensor

tensor_1d = torch.tensor([1, 2, 3])  # 1D tensor

tensor_2d = torch.tensor([[1, 2],
                          [3, 4]])  # 2D tensor

tensor_3d = torch.tensor([[[1, 2], [3, 4]],
                          [[5, 6], [7, 8]]])  # 3D tensor
```

Let's check the shapes of the tensors defined above. We can use `.shape` attribute of the tensor to access it.

```python
print("Is 0D tensor? : ", tensor_0d.shape)
print("Is 1D tensor? : ", tensor_1d.shape)
print("Is 2D tensor? : ", tensor_2d.shape)
print("Is 3D tensor? : ", tensor_3d.shape)
```

We can use `.reshape` or `.view` (the more common command for reshaping tensors in PyTorch) to reshape the tensor to a new dimension:

.view() requires the original data to be contiguous and will fail if it isn’t,
whereas .reshape() will work regardless, copying the data if necessary to ensure the desired shape.

What does "-1" do?

The "-1" in tensor reshaping is a placeholder that automatically infers the correct dimension size based on the total number of elements in the tensor and the other specified dimensions.

So, in `view((1, -1))`: 1 row and as many columns as needed to maintain the same number of elements which is 4 for this example.

<div class="alert alert-block alert-warning">
Also, pay attention the difference between a 1D tensor and 2D tensor with the same elements.
</div>

```python
print("Reshape to a 2D tensor:")
print("-----------------------")
new_shape_tensor = tensor_2d.view((1, -1))

print(new_shape_tensor)
print("Shape: ", new_shape_tensor.shape, end="\n\n")

print("Reshape to a 1D tensor:")
print("-----------------------")
new_shape_tensor = tensor_2d.view((-1))

print(new_shape_tensor)
print("Shape: ", new_shape_tensor.shape, end="\n\n")
```

### Computation Graph and Automatic Differentiation

PyTorch builds a computation graph in the background, and we can use this to calculate gradients to train the model and update the model parameters.

If we want to activate the backpropagation, we need to ensure that the `requires_grad` attribute must be set to True which is set by default.

PyTorch’s autograd engine constructs a computational graph in the background by tracking every operation performed on tensors. 
Then, calling the grad function, we can compute the gradient of the loss concerning the model parameters.

Below, we first calculate a basic arithmetic calculation as forward calculation. Then, we trigger the gradient calculation with respect to loss using `grad` method.

```python
import torch.nn.functional as F
from torch.autograd import grad

y = torch.tensor([5.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([4.4], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)
z = x1 * w1 + b
print("z: ", z)

a = torch.relu(z)  # rectified linear unit (ReLU) activation function
print("a: ", a)

loss = F.mse_loss(a, y)  # mean squared error loss
print("loss: ", loss)

grad_L_w1 = grad(loss, w1, retain_graph=True)
grad_L_b = grad(loss, b, retain_graph=True)

print("grad_L_w1: ", grad_L_w1)
print("grad_L_b: ", grad_L_b)
```

PyTorch will compute the gradients of all the leaf nodes in the graph using `backward` method, which will be stored via the tensors’ .grad attributes.

```python
loss.backward()
print(w1.grad)
print(b.grad)
```

### Multilayer Neural Networks

* While using multilayer neural networks and defining out custom network architecture we can inherit `torch.nn.Module` class which provides many useful functionalities.
For instance, it allows us to encapsulate layers and operations and keep track of the model’s parameters.
* Network structure must be defined in `__init__` constructor of the class.
* We define how the model input and the layers interact with each other in `forward` method. This is called as forward calculation.
* For backward calculations, we use `backward` method of the class but we don't need to implement it as it will automatically be calculated using the loss function and PyTorch's built-in computational graph. So, we wil just call the `backward` function.

```python
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(
            # 1st hidden layer
            torch.nn.Linear(num_inputs, 30),  # applies linear or fully connected layer
            torch.nn.ReLU(),  # applies ReLU (rectified linear unit)

            # 2nd hidden layer
            torch.nn.Linear(30, 20),
            torch.nn.ReLU(),

            # output layer
            torch.nn.Linear(20, num_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)  # input X is used in the forward pass calculations of the network
        return logits  # logits is a tensor with the shape of [20, num_outputs]
```

```python
model = NeuralNetwork(50, 3)

print("Summary of the model structure: ")
print("--------------------------------")
print(model)
```

Total number of trainable parameters of the model: 

Each parameter for which `requires_grad=True` counts as a trainable parameter and will be updated during training

```python
num_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print("Total number of trainable model parameters:", num_params)
```

```python
print("Layer's first weight matrix which is the first linear layer:")
print(model.layers[0].weight)
print("")

print("Shape of the linear layer:")
print(model.layers[0].weight.shape)
```

The shape of the weight vector of the linear layer is [30, 50] as it is expected.

```python
print("Bias vector:")
print(model.layers[0].bias)
print("")

print("Shape of the bias vector:")
print(model.layers[0].bias.shape)
```

The shape of the bias vector of the linear layer is [30] as it is expected.

When we look at the weight and bias vector, we can notice that `requires_grad` is set to True. This assignment is done by default.

We can also notice that the values in weight and bias tensors are small random numbers.
In deep learning, initializing model weights with small random numbers is desired to break symmetry during training. 
Otherwise, the nodes would be performing the same operations and updates during backpropagation, which would not allow the network to learn complex mappings from inputs to outputs.

Nevertheless, we would like to reproduce the weight values in most of the applications during development. Hence, we can easily notice if we made any mistakes. To that attempt, we use `torch.manual_seed` as the following.

```python
torch.manual_seed(123)
model = NeuralNetwork(50, 3)
print(model.layers[0].weight)
```

```python
torch.manual_seed(123)
X = torch.rand((1, 50))  # input to the model
out = model(X)  # model automatically runs the forward method and returns the output
print(out)
print("Shape of output: ", out.shape)
```

As the model is created with 50 inputs and 3 outputs, the output from the network must have 3 columns. Also, the input to the model has 1 row so the output tensor has a shape of [1, 3].

Here, `grad_fn=<AddmmBackward0>` represents the last-used function to compute a variable in the computational graph. 
In particular, `grad_fn=<AddmmBackward0>` means that the tensor we are inspecting was created via a matrix multiplication and addition operation.
PyTorch will use this information when it computes gradients during backpropagation. 
The `<AddmmBackward0>` part of `grad_fn=<AddmmBackward0>` specifies the operation performed. 
In this case, it is an `Addmm` operation. `Addmm` stands for matrix multiplication (mm) followed by an addition (Add).

Note that if we just want to use a network without training or backpropagation, constructing this computational graph for backpropagation can be wasteful as it performs unnecessary computations and consumes additional memory.

So, when we use a model for inference (for instance, making predictions) rather than training, the best practice is to use the `torch.no_grad()` context manager.

```python
with torch.no_grad():
    out = model(X)

print(out)
```

In PyTorch, it’s common practice to code models such that they return the outputs of the last layer (logits) without passing them to a nonlinear activation function.

That’s because PyTorch’s commonly used loss functions combine the softmax (or sigmoid for binary classification) operation with the negative log-likelihood loss in a single class. The reason for this is numerical efficiency and stability. So, if we want to compute class-membership probabilities for our predictions, we have to call the softmax function explicitly.

```python
with torch.no_grad():
    out = torch.softmax(model(X), dim=1)

print(out)
```

These values can be interpreted as probability values and they sum up to 1.0.

### PyTorch Data Loaders

* PyTorch implements a Dataset and a DataLoader class. 
    * The Dataset class is used to instantiate objects that define how each data record is loaded. 
    * The DataLoader handles how the data is shuffled and assembled into batches.

Example use case:

```python
# Train data
X_train = torch.tensor([
    [-1.4, 3.4],
    [-0.9, 2.9],
    [-0.5, 2.6],
    [2.3, -1.1],
    [2.7, -1.5]
])
y_train = torch.tensor([0, 0, 0, 1, 1])

# Test data
X_test = torch.tensor([
    [-0.8, 2.8],
    [2.6, -1.6],
])
y_test = torch.tensor([0, 1])

display("X_train: ", X_train)
display("y_train: ", y_train)
display("X_test: ", X_test)
display("y_test: ", y_test)
```

Let's create a toy dataset by inheriting PyTorch's Dataset class.

The purpose of this custom ToyDataset class is to instantiate a PyTorch DataLoader.

```python
from torch.utils.data import Dataset

class ToyDataset(Dataset):
    def __init__(self, X, y):
        self.features = X
        self.labels = y

    def __getitem__(self, index):  # indexing from dataset
        one_x = self.features[index]
        one_y = self.labels[index]
        return one_x, one_y
    
    def __len__(self):  # length of the dataset
        return self.labels.shape[0]
    

train_ds = ToyDataset(X_train, y_train)
test_ds = ToyDataset(X_test, y_test)
```

Now, we can use PyTorch's DataLoader class to sample from the toy dataset.

```python
from torch.utils.data import DataLoader

train_loader = DataLoader(
    dataset=train_ds,
    batch_size=2,
    shuffle=True,
    num_workers=0,  # the number of background processes, crucial for parallelizing data loading and preprocessing (0 = main process, not in separate worker process)
    drop_last=True  # when the total number of rows is not evenly divisible by 2, the remaining row will be dropped from the dataset.
)

test_loader = DataLoader(
    dataset=test_ds,
    batch_size=2,
    shuffle=False, # we will not shuffle the test data
    num_workers=0,  # the number of background processes
)
```

Now, we can iterate over the data loaders

```python
torch.manual_seed(123)  # enables the exact same shuffling order of training examples
for idx, (x, y) in enumerate(train_loader):
    print(f"Batch {idx+1}: ")
    print("----------------")
    display(x)
    display(y)
```

Notes about `num_workers`:

if you are working with tiny datasets or interactive environments such as Jupyter notebooks,
increasing num_workers may not provide any noticeable speedup.
It may, in fact, lead to some problems. One potential problem is the overhead of spinning up multiple worker processes,
which could take longer than the actual data loading when your dataset is small.

Furthermore, for Jupyter notebooks, setting num_workers to greater than 0 can sometimes lead to problems
related to the sharing of resources between different processes, resulting in errors or notebook crashes.

Setting num_workers=4 usually leads to optimal performance on many real-world datasets.

### Training Loop

Let's define a network architecture useful for classification task

```python
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(
            # 1st hidden layer
            torch.nn.Linear(num_inputs, 30),
            torch.nn.ReLU(),
            # 2nd hidden layer
            torch.nn.Linear(30, 20),
            torch.nn.ReLU(),
            # output layer
            torch.nn.Linear(20, num_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits
```

```python
import torch.nn.functional as F
torch.manual_seed(123)

model = NeuralNetwork(num_inputs=2, num_outputs=2)
optimizer = torch.optim.SGD(
    model.parameters(), lr=0.5  # model parameters are given as input to the optimizer as well as learning rate
)

num_epochs = 3
for epoch in range(num_epochs):
    model.train()  # set the model to training mode

    for batch_idx, (features, labels) in enumerate(train_loader):
        logits = model(features)
        loss = F.cross_entropy(logits, labels)  # cross entropy loss

        optimizer.zero_grad()  # Sets the gradients from the previous round to 0 to prevent unintended gradient accumulation
        loss.backward()  # Computes the gradients
        optimizer.step()  # The optimizer uses the gradients to update the model parameters
        # To prevent undesired gradient accumulation, it is important to include optimizer.zero_grad() call in each update round to reset the gradients to 0.
        # Otherwise, the gradients will accumulate, which may be undesired.

        ### LOGGING
        print(
            f"Epoch: {epoch+1:03d}/{num_epochs:03d}"
            f" | Batch {batch_idx:03d}/{len(train_loader):03d}"
            f" | Train Loss: {loss:.2f}".lstrip()
        )
    
    model.eval()  # set the model to evaluation mode
    # model.eval() is necessary for components that behave differently during training and inference,
    # such as dropout or batch normalization layers. Since we don’t have dropout or other components
    # in our NeuralNetwork class that are affected by these settings, using model.train() and model.eval()
    # is redundant in our preceding code.
    with torch.no_grad():
        outputs = model(X_train)
    print(outputs)
```

Number of parameters in the network:

```python
num_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print("Total number of trainable model parameters:", num_params)
```

```python
torch.set_printoptions(sci_mode=False)  # disables the scientific notation
probas = torch.softmax(outputs, dim=1)
print(probas)
```

Let’s consider the first row in the preceding code output. Here, the first value, first row and first column, means that the training example has a 99.96% probability of belonging to class 0 and a 0.04% probability of belonging to class 1.

We can convert these values into class label predictions using PyTorch’s argmax function,
which returns the index position of the highest value in each row if we set dim=1
(setting dim=0 would return the highest value in each column instead):

```python
predictions = torch.argmax(probas, dim=1)

print(predictions)
```

<div class="alert alert-block alert-warning">
Note that it is unnecessary to compute softmax probabilities to obtain the class labels.
We could also apply the argmax function to the logits (outputs) directly:
</div>

```python
predictions = torch.argmax(outputs, dim=1)
print(predictions)
```

Let's finally check the predictions with target values.

```python
predictions == y_train
```

Compute the accuracy:

```python
def compute_accuracy(model, dataloader):
    model = model.eval()
    correct = 0.0
    total_examples = 0

    for idx, (features, labels) in enumerate(dataloader):
        with torch.no_grad():
            logits = model(features)
            predictions = torch.argmax(logits, dim=1)
            compare = labels == predictions
            correct += torch.sum(compare)
            total_examples += len(compare)

            return (correct / total_examples).item()  # .item() returns the value of the tensor as a Python float.
        
print("Train accuracy: ", compute_accuracy(model, train_loader))
print("Test accuracy: ", compute_accuracy(model, test_loader))
```

We can write the training loop as the following. It will be generic for the systems whether GPU is available on the device or the device is only a CPU device.

```python
import torch.nn.functional as F
torch.manual_seed(123)

model = NeuralNetwork(num_inputs=2, num_outputs=2).to(device)  # model sent to device (GPU if available else CPU)
optimizer = torch.optim.SGD(
    model.parameters(), lr=0.5  # model parameters are given as input to the optimizer as well as learning rate
)

num_epochs = 3
for epoch in range(num_epochs):
    model.train()  # set the model to training mode

    for batch_idx, (features, labels) in enumerate(train_loader):
        features, labels = features.to(device), labels.to(device)  # features and labels sent to device (GPU if available else CPU)
        logits = model(features)
        loss = F.cross_entropy(logits, labels)  # cross entropy loss

        optimizer.zero_grad()  # Sets the gradients from the previous round to 0 to prevent unintended gradient accumulation
        loss.backward()  # Computes the gradients
        optimizer.step()  # The optimizer uses the gradients to update the model parameters
        # To prevent undesired gradient accumulation, it is important to include optimizer.zero_grad() call in each update round to reset the gradients to 0.
        # Otherwise, the gradients will accumulate, which may be undesired.

        ### LOGGING
        print(
            f"Epoch: {epoch+1:03d}/{num_epochs:03d}"
            f" | Batch {batch_idx:03d}/{len(train_loader):03d}"
            f" | Train Loss: {loss:.2f}".lstrip()
        )
    
    model.eval()  # set the model to evaluation mode
    # model.eval() is necessary for components that behave differently during training and inference,
    # such as dropout or batch normalization layers. Since we don’t have dropout or other components
    # in our NeuralNetwork class that are affected by these settings, using model.train() and model.eval()
    # is redundant in our preceding code.
    with torch.no_grad():
        outputs = model(X_train)
    print(outputs)
```
